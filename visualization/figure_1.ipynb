```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
sns.set_theme(style="darkgrid", rc={
    "axes.facecolor": "#f5f5f9",   
    "grid.color": "#d2d2e0",       
    "axes.edgecolor": "#8282a0",   
})
```

```python
plt.rcParams.update({
    "font.size": 12,
    "axes.titlesize": 14,
    "axes.labelsize": 13,
    "legend.fontsize": 11,
})
```

```python
latency = np.array([1, 3, 5, 7])  # x 轴
data = {
    "Bias Subset"        : [6.11, 22.74, 25.09, 27.65],
    "Sensitivity Subset" : [36.06, 43.16, 44.58, 46.54],
    "BS Subset"   : [14.52, 26.94, 29.5, 31.72],
}
```

```python
markers = ["*", "o", "s"]
colors  = sns.color_palette("deep", n_colors=len(data))
fig, ax = plt.subplots(figsize=(7, 5))
ax.set_xticks([1, 3, 5, 7])
for (label, y), m, c in zip(data.items(), markers, colors):
    ax.plot(latency, y, label=label,
            marker=m, markersize=8, linewidth=2, color=c)

ax.set_xlabel("Number of SCI Rounds")
ax.set_ylabel("Accuracy")
ax.set_title("Qwen2-VL-7B-SCI", pad=15)
#ax.set_yscale("log")
ax.set_ylim(0, 50) 

ax.legend(framealpha=1.0, loc="upper center", bbox_to_anchor=(0.8, 0.3))
plt.tight_layout()
plt.savefig("figure1c_1.png", format="png", bbox_inches='tight')
plt.show()
```
