```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
sns.set_theme(style="darkgrid", rc={
    "axes.facecolor": "#f5f5f9",   
    "grid.color": "#d2d2e0",       
    "axes.edgecolor": "#8282a0",   
})
```

```python
plt.rcParams.update({
    "font.size": 12,
    "axes.titlesize": 14,
    "axes.labelsize": 13,
    "legend.fontsize": 11,
})
```

```python
latency = np.array([1, 3, 5, 7])  # x 轴
data = {
    "B Subset"  : [5.37, 21.67, 24.91, 27.04],
    "S Subset"  : [38.1, 44.05, 47.22, 47.22],
    "BS Subset" : [10.78, 24.54, 28, 29.61],
}

markers = ["*", "o", "s"]
colors  = sns.color_palette("deep", n_colors=len(data))
fig, ax = plt.subplots(figsize=(7, 5))
ax.set_xticks([1, 3, 5, 7])
for (label, y), m, c in zip(data.items(), markers, colors):
    ax.plot(latency, y, label=label,
            marker=m, markersize=8, linewidth=2, color=c)

ax.set_xlabel("(Qwen2-VL-7B-SCI) Number of SCI Rounds")
ax.set_ylabel("(MCQ) Accuracy")
#ax.set_yscale("log")
ax.set_ylim(0, 50) 

ax.legend(framealpha=1.0, loc="upper center", bbox_to_anchor=(0.8, 0.3))
plt.tight_layout()
plt.savefig("figure4_a1.png", format="png", bbox_inches='tight')
plt.show()
```
